## 1. Introducción y Contexto del Negocio

### 1.1 Descripción de la Empresa

Home Credit es una institución financiera internacional enfocada en proporcionar servicios de préstamos responsables a personas con historial crediticio limitado o inexistente. La empresa opera en múltiples países y se especializa en:

- **Inclusión financiera:** Brindar acceso a servicios crediticios a poblaciones desatendidas
- **Préstamos al consumo:** Créditos para compras de bienes y servicios
- **Microfinanzas:** Préstamos pequeños para necesidades personales
- **Evaluación alternativa de riesgo:** Uso de datos no tradicionales para evaluar solvencia

### 1.2 Problemática del Negocio

La evaluación tradicional del riesgo crediticio presenta varios desafíos:

1. **Exclusión financiera:** Muchas personas sin historial crediticio formal quedan excluidas del sistema financiero
2. **Riesgo de default elevado:** La falta de información tradicional aumenta la incertidumbre sobre la capacidad de pago
3. **Costos operativos altos:** La evaluación manual de solicitudes es costosa y propensa a errores
4. **Decisiones subjetivas:** Las evaluaciones humanas pueden ser inconsistentes y sesgadas
5. **Tiempo de respuesta lento:** Los procesos manuales retrasan las aprobaciones de crédito

### 1.3 Oportunidad de Machine Learning

El uso de técnicas avanzadas de Machine Learning permite:

- Evaluar automáticamente el riesgo crediticio usando múltiples fuentes de datos
- Identificar patrones complejos no detectables mediante análisis tradicional
- Reducir el sesgo humano en las decisiones crediticias
- Acelerar el proceso de evaluación de solicitudes
- Mejorar la precisión en la predicción de defaults

## 3. Objetivos de Machine Learning

### 3.1 Objetivos de Clasificación

#### Objetivo 1: Predicción de Default (TARGET Principal)
- **Variable objetivo:** TARGET (0 = No default, 1 = Default)
- **Tipo de problema:** Clasificación binaria desbalanceada
- **Descripción:** Predecir si un cliente tendrá dificultades para pagar el préstamo
- **Métricas principales:**
  - ROC-AUC (métrica principal de competencia)
  - Precision-Recall AUC
  - F1-Score balanceado
- **Umbral de rendimiento:** ROC-AUC > 0.75

#### Objetivo 2: Segmentación de Riesgo
- **Variable objetivo:** Categorías de riesgo (Bajo, Medio, Alto, Muy Alto)
- **Tipo de problema:** Clasificación multiclase
- **Descripción:** Clasificar clientes en segmentos de riesgo para estrategias diferenciadas
- **Métricas:**
  - Accuracy ponderada
  - Cohen's Kappa
  - Matriz de confusión multiclase

#### Objetivo 3: Predicción de Mora Temprana
- **Variable objetivo:** Mora en primeros 60 días (derivada)
- **Tipo de problema:** Clasificación binaria
- **Descripción:** Identificar clientes con alta probabilidad de mora temprana
- **Métricas:**
  - Recall (sensibilidad)
  - Precision
  - Early Warning Score

### 3.2 Objetivos de Regresión

#### Objetivo 1: Estimación de Monto de Crédito Óptimo
- **Variable objetivo:** AMT_CREDIT (monto del crédito)
- **Tipo de problema:** Regresión
- **Descripción:** Predecir el monto óptimo de crédito según el perfil del cliente
- **Métricas:**
  - RMSE (Root Mean Square Error)
  - MAE (Mean Absolute Error)
  - R² ajustado
  - MAPE (Mean Absolute Percentage Error)

#### Objetivo 2: Predicción de Pérdida Esperada (Expected Loss)
- **Variable objetivo:** Pérdida monetaria esperada (calculada)
- **Tipo de problema:** Regresión
- **Descripción:** Estimar la pérdida financiera esperada por cliente
- **Fórmula:** EL = PD × EAD × LGD
  - PD: Probability of Default
  - EAD: Exposure at Default
  - LGD: Loss Given Default
- **Métricas:**
  - MAE ponderado por monto
  - Percentile accuracy
  - Economic value metrics

#### Objetivo 3: Estimación de Capacidad de Pago
- **Variable objetivo:** AMT_ANNUITY (anualidad del préstamo)
- **Tipo de problema:** Regresión
- **Descripción:** Predecir la capacidad máxima de pago mensual del cliente
- **Métricas:**
  - RMSE
  - Ratio de cobertura de deuda
  - Percentile errors

### 3.3 Objetivos Combinados (Ensemble)

#### Score de Crédito Integral
- **Combinación de:** Probabilidad de default + Pérdida esperada + Capacidad de pago
- **Tipo:** Score compuesto (0-1000)
- **Uso:** Decisión integral de aprobación y condiciones del préstamo

## Anexo B: Métricas de Evaluación Detalladas

### Métricas para Clasificación

| Métrica | Fórmula | Interpretación | Umbral Objetivo |
|---------|---------|----------------|----------------|
| ROC-AUC | Area Under ROC Curve | Capacidad de discriminación | > 0.75 |
| Precision | TP / (TP + FP) | Exactitud en predicciones positivas | > 0.30 |
| Recall | TP / (TP + FN) | Cobertura de casos positivos | > 0.70 |
| F1-Score | 2 × (Precision × Recall) / (Precision + Recall) | Balance precision-recall | > 0.40 |
| Cohen's Kappa | (Po - Pe) / (1 - Pe) | Concordancia ajustada por azar | > 0.50 |

### Métricas para Regresión

| Métrica | Fórmula | Interpretación | Umbral Objetivo |
|---------|---------|----------------|----------------|
| RMSE | √(Σ(y_pred - y_true)² / n) | Error promedio absoluto | < 15% del promedio |
| MAE | Σ|y_pred - y_true| / n | Error absoluto promedio | < 10% del promedio |
| R² | 1 - (SS_res / SS_tot) | Varianza explicada | > 0.70 |
| MAPE | 100 × Σ|y_pred - y_true| / y_true / n | Error porcentual promedio | < 20% |

## Anexo C: Glosario de Términos

- **Default:** Incumplimiento en el pago del préstamo
- **PD (Probability of Default):** Probabilidad de incumplimiento
- **EAD (Exposure at Default):** Exposición al momento del default
- **LGD (Loss Given Default):** Pérdida dado el default
- **Expected Loss:** Pérdida esperada = PD × EAD × LGD
- **ROC-AUC:** Area bajo la curva ROC, métrica de discriminación
- **CRISP-DM:** Cross-Industry Standard Process for Data Mining
- **Kedro:** Framework de Python para pipelines de datos reproducibles
- **Feature Engineering:** Creación de variables derivadas
- **Overfitting:** Sobreajuste del modelo a datos de entrenamiento
- **Cross-validation:** Validación cruzada para evaluar generalización
- **SMOTE:** Synthetic Minority Over-sampling Technique
